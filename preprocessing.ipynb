```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
from sklearn.feature_selection import SelectKBest, f_classif


# 1. Dataset Generation
data = {
    "Age": [25, 30, np.nan, 35, 40, 22, 29, np.nan, 45, 50],
    "Salary": [50000, 60000, 52000, np.nan, 80000, 48000, 58000, 62000, 90000, 110000],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "IT", "Sales", "HR", "IT", "Sales"],
    "Purchased": ["No", "Yes", "No", "Yes", "No", "Yes", "No", "Yes", "Yes", "No"]
}

df = pd.DataFrame(data)
print("--- Raw Dataset ---")
print(df)

# 2. Handling Missing Values (Imputation)
df["Age"].fillna(df["Age"].mean(), inplace=True)
df["Salary"].fillna(df["Salary"].median(), inplace=True)

# 3. Categorical Encoding (One-Hot Encoding)
df = pd.get_dummies(df, columns=["Department"], drop_first=True)

# Map binary target variable
df["Purchased"] = df["Purchased"].map({"No": 0, "Yes": 1})

# 4. Feature Normalization (Min-Max Scaling)
scaler = MinMaxScaler()
df[["Age", "Salary"]] = scaler.fit_transform(df[["Age", "Salary"]])

# 5. Exploratory Data Analysis (EDA) Summary
print("\n--- Processed Dataset ---")
print(df)

print("\n--- Summary Statistics ---")
print(df.describe())

# Save cleaned dataset
df.to_csv("cleaned_data.csv", index=False)
```
